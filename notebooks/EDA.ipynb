```python
"""
Análisis exploratorio básico del conjunto de datos vehicles_us.csv.

El análisis incluye distribuciones, relaciones entre variables numéricas,
comparaciones entre variables categóricas y una prueba de correlación
entre el kilometraje y el precio.
"""

import pandas as pd
import plotly.graph_objects as go


# Cargar el conjunto de datos.
car_data = pd.read_csv("vehicles_us.csv")


# Mostrar información general del conjunto de datos.
car_data.info()


# Mostrar estadísticas descriptivas de las variables numéricas.
car_data.describe()


# Crear histograma para analizar la distribución de los precios.
fig = go.Figure(
    data=[
        go.Histogram(
            x=car_data["price"],
            nbinsx=50
        )
    ]
)

fig.update_layout(
    title="Distribución de los precios de los vehículos",
    xaxis_title="Precio",
    yaxis_title="Cantidad de vehículos"
)

fig.show()


# Crear histograma para analizar la distribución del kilometraje.
fig = go.Figure(
    data=[
        go.Histogram(
            x=car_data["odometer"],
            nbinsx=50
        )
    ]
)

fig.update_layout(
    title="Distribución del kilometraje",
    xaxis_title="Kilometraje",
    yaxis_title="Cantidad de vehículos"
)

fig.show()


# Crear gráfico de dispersión para analizar la relación entre kilometraje
# y precio.
fig = go.Figure(
    data=[
        go.Scatter(
            x=car_data["odometer"],
            y=car_data["price"],
            mode="markers",
            marker={
                "size": 5,
                "opacity": 0.5
            }
        )
    ]
)

fig.update_layout(
    title="Relación entre kilometraje y precio",
    xaxis_title="Kilometraje",
    yaxis_title="Precio"
)

fig.show()


# Seleccionar únicamente los registros que tienen datos en ambas variables.
correlation_data = car_data[["odometer", "price"]].dropna()

# Calcular el coeficiente de correlación de Pearson.
correlation = correlation_data["odometer"].corr(
    correlation_data["price"]
)

print(f"Coeficiente de correlación de Pearson: {correlation:.3f}")


# Interpretar la intensidad de la correlación.
if abs(correlation) < 0.10:
    correlation_strength = "muy débil"
elif abs(correlation) < 0.30:
    correlation_strength = "débil"
elif abs(correlation) < 0.50:
    correlation_strength = "moderada"
elif abs(correlation) < 0.70:
    correlation_strength = "considerable"
else:
    correlation_strength = "fuerte"


# Interpretar la dirección de la correlación.
if correlation > 0:
    correlation_direction = "positiva"
elif correlation < 0:
    correlation_direction = "negativa"
else:
    correlation_direction = "nula"


print(
    "Conclusión: existe una correlación "
    f"{correlation_strength} y {correlation_direction} "
    "entre el kilometraje y el precio. "
    "La correlación describe una relación lineal entre las variables, "
    "pero no implica que una variable sea causa de la otra."
)


# Crear gráfico de dispersión para analizar la relación entre el año
# del modelo y el precio.
fig = go.Figure(
    data=[
        go.Scatter(
            x=car_data["model_year"],
            y=car_data["price"],
            mode="markers",
            marker={
                "size": 5,
                "opacity": 0.5
            }
        )
    ]
)

fig.update_layout(
    title="Relación entre año del modelo y precio",
    xaxis_title="Año del modelo",
    yaxis_title="Precio"
)

fig.show()


# Contar la cantidad de vehículos por tipo.
type_counts = (
    car_data["type"]
    .value_counts()
    .sort_values(ascending=False)
)

# Crear gráfico de barras para comparar los tipos de vehículos.
fig = go.Figure(
    data=[
        go.Bar(
            x=type_counts.index,
            y=type_counts.values
        )
    ]
)

fig.update_layout(
    title="Cantidad de vehículos por tipo",
    xaxis_title="Tipo de vehículo",
    yaxis_title="Cantidad de anuncios"
)

fig.show()


# Contar la cantidad de vehículos según su condición.
condition_counts = (
    car_data["condition"]
    .value_counts()
    .sort_values(ascending=False)
)

# Crear gráfico de barras para comparar las condiciones.
fig = go.Figure(
    data=[
        go.Bar(
            x=condition_counts.index,
            y=condition_counts.values
        )
    ]
)

fig.update_layout(
    title="Cantidad de vehículos por condición",
    xaxis_title="Condición",
    yaxis_title="Cantidad de anuncios"
)

fig.show()


# Calcular el precio promedio para cada condición.
condition_price = (
    car_data.groupby("condition", as_index=False)["price"]
    .mean()
    .sort_values("price", ascending=False)
)

# Crear gráfico de barras para comparar el precio promedio por condición.
fig = go.Figure(
    data=[
        go.Bar(
            x=condition_price["condition"],
            y=condition_price["price"]
        )
    ]
)

fig.update_layout(
    title="Precio promedio según condición del vehículo",
    xaxis_title="Condición",
    yaxis_title="Precio promedio"
)

fig.show()


# Contar la cantidad de vehículos según el tipo de combustible.
fuel_counts = (
    car_data["fuel"]
    .value_counts()
    .sort_values(ascending=False)
)

# Crear gráfico de barras para comparar los tipos de combustible.
fig = go.Figure(
    data=[
        go.Bar(
            x=fuel_counts.index,
            y=fuel_counts.values
        )
    ]
)

fig.update_layout(
    title="Cantidad de vehículos por tipo de combustible",
    xaxis_title="Combustible",
    yaxis_title="Cantidad de anuncios"
)

fig.show()


# Crear un boxplot para comparar la distribución de precios
# entre los diferentes tipos de vehículos.
fig = go.Figure()

for vehicle_type in car_data["type"].dropna().unique():
    price_values = car_data.loc[
        car_data["type"] == vehicle_type,
        "price"
    ]

    fig.add_trace(
        go.Box(
            y=price_values,
            name=vehicle_type
        )
    )

fig.update_layout(
    title="Distribución del precio según tipo de vehículo",
    xaxis_title="Tipo de vehículo",
    yaxis_title="Precio"
)

fig.show()


# Crear un boxplot para comparar la distribución de precios
# entre los diferentes tipos de combustible.
fig = go.Figure()

for fuel_type in car_data["fuel"].dropna().unique():
    price_values = car_data.loc[
        car_data["fuel"] == fuel_type,
        "price"
    ]

    fig.add_trace(
        go.Box(
            y=price_values,
            name=fuel_type
        )
    )

fig.update_layout(
    title="Distribución del precio según combustible",
    xaxis_title="Combustible",
    yaxis_title="Precio"
)

fig.show()


# Calcular la cantidad y el porcentaje de valores ausentes por columna.
missing_values = pd.DataFrame(
    {
        "missing_values": car_data.isna().sum(),
        "missing_percentage": car_data.isna().mean() * 100
    }
).sort_values(
    "missing_percentage",
    ascending=False
)

# Mostrar el resumen de valores ausentes.
print("Resumen de valores ausentes:")
print(missing_values)
```
